```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
np.random.seed(42)

blood_pressure = 2 * np.random.rand(100, 1)

inflammation_level = 4 + 3 * blood_pressure + np.random.randn(100, 1)

```

```python
#nasumični su brojevi, dva vektora
```

```python

data = pd.DataFrame({

'Blood Pressure (mmHg)': blood_pressure.flatten(),

'Inflammation Level (Biomarker)': inflammation_level.flatten()

})


print(data.head())
```

```python
#train test split- odvajmo podatke
```

```python
bp_train, bp_test, infl_train, infl_test= train_test_split(blood_pressure, inflammation_level,
                                                           test_size=0.2, random_state=42)
```

```python
#random state ili bez, svejedno je, ako je bez svaki put će biti drugačiji rezultat
```

```python
model= LinearRegression()
model.fit(bp_train, infl_train)
#kalibracija se uči, zovemo lin reg kao model, dolaze sa () uvijek
#fitamo na x i y
#ovo bijelo plavo znači da je spremljeno, da je naučeno
```

```python
 infl_pred= model.predict(bp_test)
infl_pred
```
