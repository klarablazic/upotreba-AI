```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#učitavanje podataka
```

```python
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
stupci = ['Trudnoce', 'Glukoza', 'Tlak', 'Koza', 'Insulin', 'BMI', 'Pedigre', 'Dob', 'Ishod']
df = pd.read_csv(url, names=stupci)
df
```

```python
null_counts = (df == 0).sum()
print(null_counts)
```

```python
cols_fix = [ 'Glukoza', 'Tlak', 'Koza', 'Insulin', 'BMI']
df[cols_fix] = df[cols_fix].replace(0, np.nan)
df.fillna(df.median(), inplace=True)
df
```

```python
null_counts = (df == 0).sum()
print(null_counts)
```

```python
df.describe()
```

```python
#učitamo dataset da vidimo kako izgleda
df.head(5)
```

```python
#odabiranje x i y varijabli, imamo 8 x varijabli te 1 y
```

```python
X = df.drop('Ishod', axis=1)
# koristimo svih 8 nezavisnih varijabli
y = df['Ishod']
```

```python
X
#sve nezavisne varijable
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
#treniranje modela
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
#standardizacija podataka
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
#kreiranje i testiranje LDA modela
```

```python
lda_model = LinearDiscriminantAnalysis()
lda_model.fit(X_train_scaled, y_train)
```

```python
y_pred = lda_model.predict(X_test_scaled)
y_pred
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
accuracy_score(y_test, y_pred)*100
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Zdravi', 'Dijabetes'])
disp.plot(cmap='Reds', ax=ax)
plt.title('Matrica zabune: LDA Klasifikacija')
plt.show()
```

```python
X_lda = lda_model.transform(X_test)
```

```python
plot_df = pd.DataFrame({'LD1': X_lda.flatten(), 'Status': y_test})
plot_df['Status'] = plot_df['Status'].map({0: 'Zdravi', 1: 'Dijabetes'})
```

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(data=plot_df, x='LD1', hue='Status', fill=True, common_norm=False, palette='viridis', alpha=0.5)

plt.title('LDA: Razdvajanje klasa na temelju prve diskriminante (LD1)')
plt.xlabel('x')
plt.ylabel('Gustoća (Density)')
plt.legend()
plt.show()
```

```python
coeffs = pd.DataFrame({'Značajka': X.columns, 'Težina': lda_model.coef_[0]})
coeffs = coeffs.sort_values(by='Težina', ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(data=coeffs, x='Težina', y='Značajka', palette='magma')
plt.title('Važnost značajki u LDA modelu')
plt.show()
```
